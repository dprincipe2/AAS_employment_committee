## Making the figures 

* Each function (figure) below is meant to be run with a start and end date in mind. This date uses the 'deadline' column of the table so we can dispay the number of jobs between two dates. This is helpful if we want to show, for example, job season (~aug-dec) in 2024 versus 2025.
* At the moment these are all count plots. One future thing we can do is convert some to line plots to show number of jobs as a function of time.
* Each function is very similar but I chose to make them all individual functions so users can modify them individually. For example, some plots might require different filtering, notes on the various axes or formatting concerns. Otherwise ~90% of each function is identical. We might also want to publish identical plots over time which make it easier to have a seperate function for each. 

```python
def make_fig_1(table_data, start_date, end_date, filename_root, file_type = 'png'):
    """
    This function makes Fig 1: A bar plot showing the number of job listings by position category (e.g., postdoc, grad student, faculty, etc).

    table_data -- astropy Table object based on jobs register data file
    start_date, end date -- interval during which you want to plot data using application deadline (ISO date format e.g., '2025-01-24')
    filename_root -- string for naming the output png file.
    plot_type -- file format for the plot
    """


    #check if input time is within bounds of table min/max dates. If not, adjust time to match min/max dates in table.
    if start_date < min(table_data['deadline']):
        date1 = min(table_data['deadline'])
    else:
        date1 = start_date
    
    if end_date > max(table_data['deadline']):
        date2 = max(table_data['deadline'])
    else:
        date2 = end_date
    
    #filter the table based on date -- format is 2024-04-28
    time_table = table_data[ (table_data['deadline'] >= date1) & (table_data['deadline'] <= date2) ] 

    table_date1 = min(time_table['deadline']) #store the min/max dates actually used in generating the plot for printing the plot title
    table_date2 = max(time_table['deadline'])


    #create the plot
    fig1, ax1 = plt.subplots(1,1)

    sns.countplot(data=time_table, y = 'category') #change y to x if you want vertical plot

    ax1.set_xlabel('Number of JR Advertisements')
    ax1.set_ylabel('') #removes the awkward category label

    ax1.set_title(f'Job Advertisments with Deadlines from {table_date1} to {table_date2} ')

    fig1.subplots_adjust(left=0.3) #necessary for fitting the long strings on left side of plot?
    fig1.set_size_inches(12,4)
    fig1.savefig(f'fig1_{filename_root}.png') #save the figure

    return()


make_fig_1(table_data = job_data, start_date = '2024-05-30', end_date = '2025-08-01', filename_root = 'job_category', file_type = 'png') #generates the figure
```

```python
def make_fig_5(table_data, start_date, end_date, filename_root, file_type = 'png', min_job_postings = 2):
    """
    This function makes Fig 5: A bar plot showing the number of job listings by country above some minimum threshold (default = 2).

    table_data -- astropy Table object based on jobs register data file
    start_date, end date -- interval during which you want to plot data using application deadline (ISO date format e.g., '2025-01-24')
    filename_root -- string for naming the output png file.
    plot_type -- file format for the plot
    min_job_postings -- the minumum number of job postings a country must have to be included in the plot ** note this is considered AFTER the time filter is done
    """


    #check if input time is within bounds of table min/max dates. If not, adjust time to match min/max dates in table.
    if start_date < min(table_data['deadline']):
        date1 = min(table_data['deadline'])
    else:
        date1 = start_date
    
    if end_date > max(table_data['deadline']):
        date2 = max(table_data['deadline'])
    else:
        date2 = end_date
    
    #filter the table based on date -- format is 2024-04-28
    time_table = table_data[ (table_data['deadline'] >= date1) & (table_data['deadline'] <= date2) ]
    
    table_date1 = min(time_table['deadline']) #store the min/max dates actually used in generating the plot for printing the plot title
    table_date2 = max(time_table['deadline'])

    minimum_jobs_country = time_table[time_table['country_short'].map(time_table['country_short'].value_counts()) >= min_job_postings ] #since most countries only have 1 job listing, this table KEEPS countries with more than X job postings between the provided dates.
                                                    
    #create the plot
    fig5, ax5 = plt.subplots(1,1)
    #sns.countplot(data=time_table, y = 'country_short') #change y to x if you want vertical plot
    sns.countplot(data=minimum_jobs_country, y = 'country_short') #this requires a minimum number of jobs listings between the date range to be included in the plot because most countries have only 1 job listing.
    
    ax5.set_xlabel('Number of JR Advertisements')
    ax5.set_ylabel(f'Countries with at least {min_job_postings} JR advertisments')
    ax5.set_title(f'Job Advertisments with Deadlines from {date1} to {date2} ')

    fig5.subplots_adjust(left=0.1) #necessary for fitting the long strings on left side of plot?
    fig5.set_size_inches(12,4)
    fig5.savefig(f'fig5_{filename_root}.png') #save the figure

    return()


make_fig_5(table_data = job_data, start_date = '2024-05-01', end_date = '2026-12-12', filename_root = 'country', file_type = 'png', min_job_postings = 2) #generates the figure
```
